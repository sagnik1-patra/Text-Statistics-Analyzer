```python
from pathlib import Path
from collections import Counter, defaultdict
from urllib.parse import urlparse
from itertools import chain
import re, json, math, zipfile
import numpy as np
import pandas as pd
import matplotlib
# Jupyter uses its active inline backend; do not force 'Agg'.
import matplotlib.pyplot as plt
from scipy.io import mmread
from reportlab.platypus import SimpleDocTemplate, Paragraph, Spacer, Table, TableStyle, PageBreak
from reportlab.lib.styles import getSampleStyleSheet
from reportlab.lib import colors

DEFAULT = Path(r'C:\Users\sagni\Downloads\Text Statistics Analyzer')
# When running the supplied notebook in another environment, fall back to mounted files.
if not (DEFAULT / 'bbcsport.mtx').exists() and (Path('/mnt/data') / 'bbcsport.mtx').exists():
    DEFAULT = Path('/mnt/data')
COLORS = ['#2563eb','#059669','#ea580c','#7c3aed','#dc2626']

def read_lines(path):
    """Read text with fallback encodings."""
    for enc in ('utf-8-sig','ascii','cp1252','latin-1'):
        try: return path.read_text(encoding=enc).splitlines()
        except UnicodeError: pass
    raise UnicodeError(f'Cannot decode {path}')

def load_data(root):
    """Read BBC Sport term-document sparse matrix and metadata."""
    terms = [s.strip() for s in read_lines(root/'bbcsport.terms')]
    docs = [s.strip() for s in read_lines(root/'bbcsport.docs')]
    mat = mmread(str(root/'bbcsport.mtx')).tocsr()
    if mat.shape != (len(terms),len(docs)):
        raise ValueError(f'Matrix {mat.shape} incompatible with {len(terms)} terms and {len(docs)} docs')
    lines=read_lines(root/'bbcsport.classes')
    category_names=[]; classes={}
    for line in lines:
        if line.startswith('%Clusters'):
            category_names=line.split(' ',2)[-1].replace(',',' ').split()
        elif re.match(r'^\d+\s+\d+\s*$',line):
            a,b=map(int,line.split()); classes[a]=b
    # .classes uses zero-based IDs; Matrix Market uses one-based indices.
    categories=[category_names[classes[i]] if i in classes and classes[i]<len(category_names) else docs[i].split('.')[0] for i in range(len(docs))]
    return terms,docs,mat,categories

def tokenize(text):
    """Extract Unicode alphabetic words, allowing internal apostrophes."""
    return re.findall(r"[^\W\d_]+(?:'[^\W\d_]+)*",text.lower(),re.UNICODE)

def extract_patterns(text):
    """Find text patterns using Python's built-in regular expressions."""
    emails=re.findall(r'(?<![\w.+-])[\w.+-]+@[\w.-]+\.[A-Za-z]{2,}\b',text)
    phones=re.findall(r'(?<!\w)(?:\+1[-.\s]?)?(?:\(\d{3}\)|\d{3})[-.\s]?\d{3}[-.\s]?\d{4}(?!\d)',text)
    normalized=[]
    for p in phones:
        digits=re.sub(r'\D','',p)
        if len(digits)==11 and digits[0]=='1': digits=digits[1:]
        if len(digits)==10: normalized.append(f'+1-{digits[:3]}-{digits[3:6]}-{digits[6:]}')
    urls=re.findall(r'https?://[^\s<>"\']+|www\.[^\s<>"\']+',text,re.I)
    urls=[u.rstrip('.,;:!?)]}') for u in urls]
    domains=Counter(urlparse(u if '://' in u else 'https://'+u).netloc.lower() for u in urls)
    dates=re.findall(r'\b(?:\d{1,2}/\d{1,2}/\d{4}|\d{1,2}-\d{1,2}-\d{4}|(?:January|February|March|April|May|June|July|August|September|October|November|December)\s+\d{1,2},?\s+\d{4})\b',text,re.I)
    currency=re.findall(r'(?<!\w)[$£€]\s?\d[\d,]*(?:\.\d{1,2})?',text)
    return {'emails':{'total':len(emails),'unique':sorted(set(emails))},'phones':{'total':len(phones),'normalized':sorted(set(normalized))},'hashtags':sorted(set(re.findall(r'(?<!\w)#[A-Za-z][\w]*',text))),'mentions':sorted(set(re.findall(r'(?<!\w)@[A-Za-z][\w]*',text))),'urls':{'total':len(urls),'unique':sorted(set(urls)),'domains':dict(domains)},'dates':dates,'currency':currency}

def plot_save(name,fn,out):
    plt.figure(figsize=(11,6)); fn(); plt.tight_layout(); plt.savefig(out/name,dpi=130,bbox_inches='tight'); plt.close()

def report(out,stats,findings):
    """Create a 2-3 page report summarizing reliable and unavailable metrics."""
    styles=getSampleStyleSheet(); story=[]
    def p(t,style='BodyText'): story.append(Paragraph(str(t),styles[style]));story.append(Spacer(1,10))
    p('BBC Sport Text Statistics Analysis','Title')
    p('Methodology and data limitations','Heading2')
    p('The supplied BBC Sport files are a preprocessed term-document count matrix, not original article text. Word counts and term statistics are measured from the matrix. Sentence, punctuation, contact details, and original character counts cannot be recovered. Any reconstructed text is synthetic and should not be treated as verbatim source.')
    rows=[['Metric','Value']]+[[str(k),str(v)] for k,v in stats.items()]
    tbl=Table(rows,colWidths=[250,210],repeatRows=1);tbl.setStyle(TableStyle([('BACKGROUND',(0,0),(-1,0),colors.HexColor('#dbeafe')),('GRID',(0,0),(-1,-1),0.4,colors.grey),('VALIGN',(0,0),(-1,-1),'TOP')]));story.append(tbl)
    story.append(PageBreak());p('Findings and observations','Heading1')
    for heading,body in findings:
        p(heading,'Heading2');p(body)
    story.append(PageBreak());p('Interpretation and recommendations','Heading1')
    p('Word frequency follows a heavy-tailed pattern; compare the rank-frequency chart with a Zipf-like reference rather than assuming an exact power law. Frequent terms can reflect general sports reporting and should not automatically be interpreted as distinctive features.')
    p('Vocabulary richness is reported as unique terms divided by observed token counts. This metric is sensitive to document length and to the preprocessing used to create the source vocabulary.')
    p('Regex challenges and limitations','Heading2')
    p('The original punctuation and context were discarded in the supplied matrix, making faithful detection of email addresses, phone numbers, dates, URLs, and sentences impossible. The program therefore runs regex only on separately provided original TXT files and marks unavailable values explicitly when these are absent. Phone normalization assumes North American ten-digit formats.')
    p('Further work','Heading2');p('Supply the original BBC articles as TXT files for exact character and sentence statistics, meaningful regex extraction, and reliable n-grams. Review extracted patterns manually to evaluate false positives.')
    SimpleDocTemplate(str(out/'analysis_report.pdf'),pagesize=(595,842),leftMargin=55,rightMargin=55,topMargin=55,bottomMargin=55).build(story)

def main(root,show=False):
    root.mkdir(parents=True,exist_ok=True);out=root/'results';out.mkdir(exist_ok=True)
    terms,docs,mat,categories=load_data(root)
    doc_mat=mat.T.tocsr();counts=np.asarray(doc_mat.sum(axis=1)).ravel().astype(int)
    term_counts=np.asarray(mat.sum(axis=1)).ravel().astype(int)
    freq=Counter({t:int(n) for t,n in zip(terms,term_counts) if n>0})
    df=pd.DataFrame({'document_id':np.arange(len(docs)),'filename':docs,'category':categories,'word_count':counts,'character_count':pd.NA,'unique_words':np.diff(doc_mat.indptr)})
    df['vocabulary_richness']=np.where(df.word_count>0,df.unique_words/df.word_count,0)
    df['character_count_status']='unavailable: original text absent'
    raw_files=sorted((root/'original_text').rglob('*.txt')) if (root/'original_text').exists() else []
    pattern_data={}; sentences=[]; all_raw=[]
    for path in raw_files:
        content='\n'.join(read_lines(path));all_raw.append(content)
        pattern_data[path.name]=extract_patterns(content)
        sentences.extend(s.strip() for s in re.split(r'(?<=[.!?])\s+',content) if s.strip())
    if raw_files:
        for i,path in enumerate(raw_files):
            content='\n'.join(read_lines(path)); ws=tokenize(content)
            df.loc[len(df)]={'document_id':f'raw_{i}','filename':path.name,'category':'original_txt','word_count':len(ws),'character_count':len(content),'unique_words':len(set(ws)),'vocabulary_richness':len(set(ws))/max(1,len(ws)),'character_count_status':'exact'}
    df.to_csv(out/'document_statistics.csv',index=False)
    (out/'extracted_patterns.json').write_text(json.dumps({'source_warning':'BBC matrix lacks original strings; extraction only performed on original_text/*.txt','documents':pattern_data},indent=2,ensure_ascii=False),encoding='utf-8')
    # One line per document, reconstructed from counts; preserve matrix terms without pretending it is original prose.
    with (out/'cleaned_processed_text.txt').open('w',encoding='utf-8') as f:
        f.write('RECONSTRUCTED TERM MULTISETS: NOT ORIGINAL ARTICLE TEXT\n')
        for i,name in enumerate(docs):
            row=doc_mat.getrow(i); words=list(chain.from_iterable([terms[j]]*int(n) for j,n in zip(row.indices,row.data)))
            f.write(f'{name}: '+ ' '.join(words)+'\n')
    vowels=set('aeiou'); word_rows=[]
    for word,n in freq.items():
        word_rows.append((word,n,len(word),word[0] if word else '',sum(c in vowels for c in word)))
    words=pd.DataFrame(word_rows,columns=['word','frequency','length','first_char','vowel_count']).set_index('word')
    words.sort_values('frequency',ascending=False).to_csv(out/'word_statistics.csv')
    words.groupby('length').agg(word_types=('frequency','size'),token_count=('frequency','sum'),mean_frequency=('frequency','mean')).to_csv(out/'group_by_word_length.csv')
    words.groupby('first_char').agg(word_types=('frequency','size'),token_count=('frequency','sum')).to_csv(out/'group_by_first_character.csv')
    words.nlargest(50,'frequency').to_csv(out/'top_50_words.csv')
    words[words.length>10].to_csv(out/'words_longer_than_10.csv')
    pd.pivot_table(words.reset_index(),index='first_char',columns='length',values='frequency',aggfunc='sum',fill_value=0).to_csv(out/'word_pivot_table.csv')
    pd.crosstab(words.first_char,words.length).to_csv(out/'word_crosstab.csv')
    category_df=df.iloc[:len(docs)].groupby('category').agg(documents=('filename','count'),tokens=('word_count','sum'),mean_richness=('vocabulary_richness','mean'))
    category_df.to_csv(out/'category_statistics.csv')
    lengths=np.array([len(t) for t in terms],dtype=int);weighted_lengths=np.repeat(lengths,term_counts)
    chars=Counter();letter_chars=Counter();punct_chars=Counter()
    for term,n in freq.items():
        for c in term:
            chars[c]+=n
            if c.isalpha():letter_chars[c]+=n
            elif not c.isalnum():punct_chars[c]+=n
    pd.DataFrame(chars.most_common(),columns=['character','frequency']).to_csv(out/'character_statistics.csv',index=False)
    arr=np.array(list(freq.values()),dtype=float)
    sentence_lengths=np.array([len(tokenize(s)) for s in sentences],dtype=int)
    stats={'documents':len(docs),'categories':len(set(categories)),'vocabulary_size':len(freq),'total_term_occurrences':int(sum(freq.values())),'mean_word_frequency':round(float(arr.mean()),3),'median_word_frequency':round(float(np.median(arr)),3),'std_word_frequency':round(float(arr.std()),3),'frequency_percentiles_25_50_75':str(np.percentile(arr,[25,50,75]).round(2).tolist()),'average_word_length':round(float(weighted_lengths.mean()),3),'vowel_consonant_ratio':round(sum(n for c,n in letter_chars.items() if c in vowels)/max(1,sum(n for c,n in letter_chars.items() if c not in vowels)),3),'original_sentence_count':len(sentences) if raw_files else 'unavailable','average_sentence_length':round(float(sentence_lengths.mean()),2) if len(sentence_lengths) else 'unavailable'}
    (out/'summary_statistics.json').write_text(json.dumps(stats,indent=2),encoding='utf-8')
    top=freq.most_common(20); names=[x for x,_ in top];values=[y for _,y in top]
    def bar_top():
        plt.barh(names[::-1],values[::-1],color=COLORS[0]);plt.title('Top 20 words');plt.xlabel('Occurrences')
        for j,v in enumerate(values[::-1]):plt.text(v,j,f' {v}',va='center',fontsize=8)
    plot_save('01_top_20_words.png',bar_top,out)
    def hist():
        plt.hist(weighted_lengths,bins=np.arange(0,max(weighted_lengths)+2)-.5,density=True,color=COLORS[1],alpha=.75,label='Observed')
        mu=weighted_lengths.mean();sd=weighted_lengths.std();x=np.linspace(0,max(weighted_lengths),250)
        if sd>0:plt.plot(x,np.exp(-.5*((x-mu)/sd)**2)/(sd*np.sqrt(2*np.pi)),color=COLORS[4],label='Normal reference')
        plt.title('Word length distribution');plt.xlabel('Length');plt.ylabel('Density');plt.legend()
    plot_save('02_word_length_histogram.png',hist,out)
    def charplot(counter,title):
        a=counter.most_common(15);plt.bar([repr(c) for c,_ in a],[n for _,n in a],color=COLORS[2]);plt.title(title);plt.xlabel('Character');plt.ylabel('Occurrences')
    plot_save('03_character_frequency.png',lambda:charplot(Counter({c:n for c,n in chars.items() if not c.isspace()}),'Top 15 characters in processed terms'),out)
    plot_save('04_letter_frequency.png',lambda:charplot(letter_chars,'Top 15 letters'),out)
    if punct_chars:plot_save('05_punctuation_frequency.png',lambda:charplot(punct_chars,'Punctuation in processed terms'),out)
    else:
        plot_save('05_punctuation_frequency.png',lambda:(plt.text(.5,.5,'Punctuation unavailable from preprocessed vocabulary',ha='center',va='center',transform=plt.gca().transAxes),plt.axis('off')),out)
    def box():
        if len(sentence_lengths):plt.boxplot(sentence_lengths,vert=True);plt.ylabel('Words per sentence')
        else:plt.text(.5,.5,'Requires original article TXT files',ha='center',transform=plt.gca().transAxes);plt.xticks([])
        plt.title('Sentence lengths (original text only)')
    plot_save('06_sentence_length_boxplot.png',box,out)
    def zipf():
        ranked=np.sort(arr)[::-1];r=np.arange(1,len(ranked)+1)
        plt.loglog(r,ranked,label='Observed',color=COLORS[0]);plt.loglog(r,ranked[0]/r,'--',label='Zipf reference (slope -1)',color=COLORS[4]);plt.xlabel('Rank (log)');plt.ylabel('Frequency (log)');plt.title('Zipf rank-frequency analysis');plt.legend()
    plot_save('07_zipf_law.png',zipf,out)
    def richness():
        vals=[df.iloc[:len(docs)].loc[df.category==c,'vocabulary_richness'].values for c in category_df.index]
        
        import inspect
        boxplot_arg = 'tick_labels' if 'tick_labels' in inspect.signature(plt.boxplot).parameters else 'labels'
        plt.boxplot(vals, **{boxplot_arg: list(category_df.index)})
        plt.ylabel('Unique terms / tokens')
        plt.title('Vocabulary richness by category')
    plot_save('08_vocabulary_richness.png',richness,out)
    def category_chart():
        category_df.tokens.plot(kind='bar',color=COLORS[:len(category_df)]);plt.title('Token counts by category');plt.ylabel('Tokens');plt.xticks(rotation=20)
    plot_save('09_category_token_comparison.png',category_chart,out)
    fig,ax=plt.subplots(2,2,figsize=(13,10))
    ax[0,0].barh(names[:10][::-1],values[:10][::-1],color=COLORS[0]);ax[0,0].set_title('Top words')
    ax[0,1].hist(weighted_lengths,bins=25,color=COLORS[1]);ax[0,1].set_title('Word lengths')
    cc=letter_chars.most_common(12);ax[1,0].bar([c for c,_ in cc],[n for _,n in cc],color=COLORS[2]);ax[1,0].set_title('Letters')
    ax[1,1].bar(category_df.index,category_df.tokens,color=COLORS[:len(category_df)]);ax[1,1].tick_params(axis='x',rotation=20);ax[1,1].set_title('Category tokens')
    fig.suptitle('BBC Sport Text Statistics Dashboard',fontsize=17);fig.tight_layout();fig.savefig(out/'10_dashboard_2x2.png',dpi=130);plt.close(fig)
    # N-grams require sequential source text. Never compute them from unordered matrix term counts.
    grams=Counter()
    for text in all_raw:
        tok=tokenize(text)
        for n in (2,3):grams.update(' '.join(tok[i:i+n]) for i in range(len(tok)-n+1))
    pd.DataFrame(grams.most_common(100),columns=['ngram','frequency']).to_csv(out/'ngrams_original_text_only.csv',index=False)
    positives={'good','great','excellent','success','win','victory','best','strong'}
    negatives={'bad','poor','loss','lose','defeat','weak','failure','injury'}
    (out/'sentiment_keywords.json').write_text(json.dumps({'positive':sum(freq[w] for w in positives),'negative':sum(freq[w] for w in negatives),'note':'Illustrative keyword counts, not validated sentiment analysis'},indent=2))
    findings=[('Frequent terms',f'The highest-frequency terms are: {", ".join(names[:10])}. These terms may reflect shared sports vocabulary rather than category-specific themes.'),('Category differences',f'Token totals vary by category: '+', '.join(f'{k}: {int(v)}' for k,v in category_df.tokens.items())+'. Compare vocabulary richness across categories with caution because document lengths vary.'),('Character-level patterns',f'The most frequent processed characters are {", ".join(repr(c) for c,_ in chars.most_common(8))}. Character counts are derived from processed vocabulary, not original article text.'),('Zipf and anomalies','The log-log frequency curve shows whether a small number of words dominate the corpus. Unusually frequent terms can arise from reporting style or from preprocessing artifacts.'),('Pattern extraction',f'Original TXT files analyzed: {len(raw_files)}. The matrix does not retain regex-matchable contact details, dates, or URLs.')]
    report(out,stats,findings)
    (out/'README.txt').write_text('BBC SPORT ANALYZER\nRun: python text_statistics_analyzer.py --root "C:\\Users\\sagni\\Downloads\\Text Statistics Analyzer"\nMatrix terms are stemmed/preprocessed; original sentences, dates and contact details cannot be recovered. Add source articles under original_text/*.txt to enable exact regex and sentence analysis.\n',encoding='utf-8')
    archive=root/'Text_Statistics_Analyzer_Results.zip'
    with zipfile.ZipFile(archive,'w',zipfile.ZIP_DEFLATED,compresslevel=9) as z:
        for f in out.rglob('*'):
            if f.is_file():z.write(f,f.relative_to(root))
        for candidate in (root/'text_statistics_analyzer_jupyter.py', root/'Text_Statistics_Analyzer_Jupyter.ipynb'):
            if candidate.is_file(): z.write(candidate,candidate.name)
    size=archive.stat().st_size/1024**2
    if size>=50:raise RuntimeError(f'Archive exceeds 50 MB: {size:.2f} MB')
    print('Done! ZIP:',archive,'size:',round(size,2),'MB')
    print(json.dumps(stats,indent=2))
    if show:
        import matplotlib.image as mpimg
        plt.figure(figsize=(12,9));plt.imshow(mpimg.imread(out/'10_dashboard_2x2.png'));plt.axis('off');plt.show()

# JUPYTER ENTRY POINT: run this cell directly, no argparse required.
main(DEFAULT, show=True)
```
